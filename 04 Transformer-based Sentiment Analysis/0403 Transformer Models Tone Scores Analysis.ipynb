# fine-tuned bert models

```python
import pandas as pd

file_paths = [
    '../00 Cleaned Dataset/02 fine-tuned bert results/econobert_speeches.csv',
    '../00 Cleaned Dataset/02 fine-tuned bert results/finbert_speeches.csv',
    '../00 Cleaned Dataset/02 fine-tuned bert results/roberta_speeches.csv'
]

dataframes = [pd.read_csv(file) for file in file_paths]

# Display summary statistics
for i, df in enumerate(dataframes):
    print(f"Summary Statistics for BERT Model {i+1}:")
    if 'n_sent' in df.columns:
        display(df['tone_mean'].describe())
    else:
        print("Column 'tone_mean' not found in this DataFrame.")
    print("-" * 30)
```
