### Standardize Speaker Column

```python
def process_speaker_data(df):
    raw_speakers = df['Speaker'].copy()

    # STAGE 1: ROLE EXTRACTION
    def extract_role_at_time(s):
        s = str(s).lower()
        # 1. Handle "delivered by" (Author is at the start)
        base_text = s.split('delivered')[0] if 'delivered' in s else s
        # 2. Handle "on behalf of" (Author is at the end)
        base_text = base_text.split('on behalf of')[-1] if 'on behalf of' in base_text else base_text
        
        if 'former' in base_text: return 'Former Governor'
        if 'oic' in base_text or 'officer-in-charge' in base_text: return 'Officer-in-Charge'
        if 'deputy governor' in base_text: return 'Deputy Governor'
        if 'governor' in base_text or 'gov.' in base_text: return 'Governor'
        return 'Other'

    df['Role'] = raw_speakers.apply(extract_role_at_time)

    # STAGE 2: SPEAKER NAME CLEANING
    def get_clean_part(s):
        s = str(s).lower()
        part = s.split('delivered')[0] if 'delivered' in s else s
        part = part.split('on behalf of')[-1] if 'on behalf of' in part else part
        part = part.split(' and ')[-1] if 'joint message' in part else part
        return part.strip()

    clean_names = raw_speakers.apply(get_clean_part)

    # Stripping Titles
    titles_pattern = r'|'.join([
        r'^bsp\s+', r'^gov\.\s+', r'^governor\s+', r'^deputy\s+governor\s+', 
        r'^officer-in-charge\s+', r'^oic\s+', r'^former\s+', r'^mbm\s+',
        r'^bangko\s+sentral\s+ng\s+pilipinas\s+governor\s+', r'^by\s+',
        r'^\.\s*', r',\s*ph\.d\.?$', r',\s*as$'
    ])
    clean_names = clean_names.str.replace(titles_pattern, '', regex=True, flags=re.IGNORECASE).str.strip()

    # Final Unified Mapping
    def finalize_author_name(name):
        n = str(name).lower()
        if 'diokno' in n: return 'Benjamin E. Diokno'
        if 'tetangco' in n: return 'Amando M. Tetangco, Jr.'
        if 'espenilla' in n: return 'Nestor A. Espenilla, Jr.'
        if 'tuano-amador' in n or 'tua?o-amador' in n: return 'Maria Almasara Cyd N. Tuano-Amador'
        if 'guinigundo' in n: return 'Diwa C. Guinigundo'
        if 'buenaventura' in n: return 'Rafael Buenaventura'
        if 'singson' in n: return 'Gabriel Singson'
        return name.title()

    df['Speaker'] = clean_names.apply(finalize_author_name)

    # STAGE 3: SURNAME EXTRACTION
    suffixes = ['jr', 'jr.', 'ph.d', 'ph.d.', 'iii', 'ii', 'fss', 'as']
    
    def extract_surname(name):
        if not name: return None
        # Replace comma with space to catch "Tetangco,Jr"
        parts = str(name).replace(',', ' ').split()
        for word in reversed(parts):
            if word.lower().strip('.') not in suffixes:
                return word
        return parts[-1]

    df['last_name'] = df['Speaker'].apply(extract_surname)
    
    return df

bsp = process_speaker_data(bsp)
print(bsp[['Speaker', 'Role', 'last_name']].drop_duplicates())
```

### Identify and Drop Nearly Identical Speeches (Cosine Similarity)

There are some speeches that are not exact duplicates but are nearly identical in content (e.g., matching >98.5%). We drop the later occurrences of these speeches for two main reasons:

1. **Data Quality / Upload Errors:** Some instances are clearly administrative errors on the BSP website (e.g., a 2002 speech re-uploaded as a 2003 speech).
2. **Information Innovation:** Even when a Governor intentionally reiterates the exact same speech to a new audience days later, the financial market has already absorbed the 'surprise' or 'information innovation' from the first delivery. Keeping the duplicate risks double-counting the market's reaction to the same policy signal.

We use Cosine Similarity based on word frequencies to identify and drop these near-duplicates.

```python
import re
import math
from collections import Counter

# Tokenization and Cosine Similarity functions
def get_words(text):
    return re.findall(r'\w+', str(text).lower())

def cosine_similarity(counter1, counter2):
    intersection = set(counter1.keys()) & set(counter2.keys())
    numerator = sum([counter1[x] * counter2[x] for x in intersection])
    
    sum1 = sum([counter1[x]**2 for x in counter1.keys()])
    sum2 = sum([counter2[x]**2 for x in counter2.keys()])
    denominator = math.sqrt(sum1) * math.sqrt(sum2)
    
    if not denominator:
        return 0.0
    return float(numerator) / denominator

# Tokenize all texts
texts = bsp['Text'].fillna('')
counters = [Counter(get_words(t)) for t in texts]

threshold = 0.985
indices_to_drop = set()

print("Finding nearly identical speeches...")
for i in range(len(counters)):
    if i in indices_to_drop: continue
    len_i = sum(counters[i].values())
    if len_i == 0: continue
    
    for j in range(i + 1, len(counters)):
        if j in indices_to_drop: continue
        len_j = sum(counters[j].values())
        if len_j == 0: continue
        
        if abs(len_i - len_j) / max(len_i, len_j) > 0.2:
            continue
            
        sim = cosine_similarity(counters[i], counters[j])
        if sim > threshold:
            print(f'Near duplicate found (sim={sim:.4f}):')
            print(f'  Keep: Index {i} - "{bsp.iloc[i]["Title"]}" ({bsp.iloc[i]["Date"]})')
            print(f'  Drop: Index {j} - "{bsp.iloc[j]["Title"]}" ({bsp.iloc[j]["Date"]})\n')
            indices_to_drop.add(j)
            
# Drop the identified near-duplicates
bsp = bsp.drop(index=list(indices_to_drop))
bsp.reset_index(drop=True, inplace=True)
print(f"Dropped {len(indices_to_drop)} nearly identical speeches. Remaining: {len(bsp)}")
```
